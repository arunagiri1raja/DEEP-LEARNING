# **PART A – VISUALIZATION OF ACTIVATION FUNCTION**

```python


import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

print("NAME: Arunagiri R\nROLL NO: 24BAD013")
```

**Create Input Values**

```python


x = np.linspace(-10, 10, 200)

print(x[:10])
```

**Sigmoid Activation Function**

```python
# Sigmoid Activation Function

y_sigmoid = tf.keras.activations.sigmoid(x)

print(y_sigmoid[:10])
```

**Tanh Activation Function**

```python
# Tanh Activation Function

y_tanh = tf.keras.activations.tanh(x)

print(y_tanh[:10])
```

**ReLU Activation Function**

```python
# ReLU Activation Function

y_relu = tf.keras.activations.relu(x)

print(y_relu[:10])
```

**Plot Activation Functions**

```python


plt.figure(figsize=(10, 6))

plt.plot(x, y_sigmoid, label="Sigmoid", linewidth=2)
plt.plot(x, y_tanh, label="Tanh", linewidth=2)
plt.plot(x, y_relu, label="ReLU", linewidth=2)

plt.title("Activation Functions")
plt.xlabel("Input")
plt.ylabel("Output")
plt.legend()
plt.grid(True)

plt.show()
```
